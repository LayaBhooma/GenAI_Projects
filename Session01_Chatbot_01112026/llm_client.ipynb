```python

from urllib import response
from openai import OpenAI
client = OpenAI(api_key=my_api_key)


prompt = input("What is your question, (or type 'exit' to quit): ") 
response = client.chat.completions.create( 

    model="gpt-5-nano", #"gpt-5-turbo", #"gpt-5", #"gpt-5-mini", #"gpt-5-nano"
    messages=[
        {"role": "system", "content": "You are a helpful assistant. Answer the user's questions concisely."},
        {"role": "user", "content": prompt}
    ]
)
print (response.choices[0].message.content)
```

```python

#Reference Data
expected_answer = "Europe"    
actual_response = response.choices[0].message.content

# Call ChatCompltions API to compare expected vs actual
#LLM AS A JUDGE
comparison_prompt = f"""
You are an expert evaluator. Compare the actual response to the expected answer and determine if they match in meaning.
Expected Answer: {expected_answer}
Actual Response: {actual_response}
Do they match in meaning? Answer with 'Yes' or 'No' and provide a brief explanation.
"""
comparison_response = client.chat.completions.create(
    model="gpt-5-nano",     
    messages=[
        {"role": "system", "content": "You are an expert evaluator."},
        {"role": "user", "content": comparison_prompt}
    ]
)   
print(f"\nIs Expected Answer similar to the actual answer: {comparison_response.choices[0].message.content}")

```
