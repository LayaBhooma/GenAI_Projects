```python
from typing import TypedDict, Optional
from langgraph.graph import StateGraph, START, END

# class MyState extends TypedDict
class MyState(TypedDict):
    # name: Optional[str]
    # age: Optional[int]
    name: str
    age: int
    grade: int

# MyState node_a(MyState state) in Java style
def node_a(state: MyState) -> MyState:
    print("Before A run")
    print(state) # state =  {"name": "", "age": 0, "grade":0}

    print("A running....")

    state["name"] = "Node A"
    state["age"] = 11
    state["grade"] = state["age"] - 5
    print("A ran")
    print(f"node_a: {state['name']} is {state['age']} years old and in grade {state['grade']}") # node_a: Node A is 11 years old and in grade 6
    return state # state = {"name": "Node A", "age": 11, "grade": 6}

def node_b(state: MyState) -> MyState:
    print("Before B run")
    print(state) # state = {"name": "Node A", "age": 11, "grade": 6}
    state["name"] = "Node B"
    state["age"] = (state["age"] or 0) + 5
    state["grade"] = state["age"] - 5
    print(f"node_b: {state['name']} is {state['age']} years old and in grade {state['grade']}") # node_b: Node B is 16 years old
    return state


builder = StateGraph(MyState)
builder.add_node("A", node_a)
builder.add_node("B", node_b)

builder.add_edge(START, "A")
builder.add_edge("A", "B")
builder.add_edge("B", END)

graph = builder.compile()

graph = builder.compile()
print(graph.get_graph().draw_ascii())
graph.invoke({})
```

```python
# initial_state={"name":"START NODE", "age":0, grade:0}

initial_state={}
result = graph.invoke(initial_state)
print(result)

# Initial state: 
state = {"name":"START NODE", "age":0, "grade": 0}

# After Node A executes:
state = {"name": "Node A", "age": 11, "grade": 6}

# After Node B executes:
state = {"name": "Node B", "age": 16, "grade": 11}

# Final state after graph execution: 
print("Final state after graph execution:")
state = {"name": "Node B", "age": 16, "grade": 11}
```
